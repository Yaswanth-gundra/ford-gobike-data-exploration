## Investigation Overview

The goal of this presentation is to communicate the key findings from the exploratory analysis of the Ford GoBike dataset. The presentation focuses on understanding trip duration patterns, identifying the most active user groups, and examining how rider characteristics influence trip duration.

The main findings are:

- Most bike trips are short in duration.
- Subscribers make the majority of trips and generally take shorter rides than customers.
- Rider age has little influence on trip duration, while user type has a stronger effect.


## Dataset Overview and Executive Summary

The Ford GoBike dataset contains information about 183,412 bike-sharing trips made in the San Francisco Bay Area during February 2019. The dataset includes trip duration, user type, gender, rider birth year, station information, and trip timestamps.

Key findings from the analysis include:

- Most trips are short and the distribution of trip duration is highly right-skewed.
- Subscribers account for the majority of bike trips.
- Customers generally take longer trips than subscribers.
- Rider age has little effect on trip duration.

```python
# import all packages and set plots to be embedded inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# suppress warnings from final output
import warnings
warnings.simplefilter("ignore")

# load in the dataset into a pandas dataframe
df = pd.read_csv('201902-fordgobike-tripdata.csv')

# Make a copy of the dataset
bike = df.copy()

# Convert start and end times to datetime format
bike['start_time'] = pd.to_datetime(bike['start_time'])
bike['end_time'] = pd.to_datetime(bike['end_time'])

# Create age column
bike['member_age'] = 2019 - bike['member_birth_year']
```

## Trip Duration Distribution

```python
plt.figure(figsize=(8,5))

plt.hist(bike['duration_sec'], bins=50)

plt.title('Distribution of Trip Duration')
plt.xlabel('Trip Duration (Seconds)')
plt.ylabel('Number of Trips')

plt.show()
```

### Observation

Most bike trips are short, resulting in a highly right-skewed distribution. Only a small number of trips have very long durations, indicating that the Ford GoBike system is mainly used for short-distance travel.

## Trips by User Type

```python
plt.figure(figsize=(6,5))

sns.countplot(data=bike, x='user_type')

plt.title('Trips by User Type')
plt.xlabel('User Type')
plt.ylabel('Number of Trips')

plt.show()
```

### Observation

Subscribers make significantly more trips than customers. This indicates that regular members are the primary users of the Ford GoBike bike-sharing system.

## Trip Duration by User Type

```python
plt.figure(figsize=(8,6))

sns.boxplot(
    data=bike,
    x='user_type',
    y='duration_sec'
)

plt.title('Trip Duration by User Type')
plt.xlabel('User Type')
plt.ylabel('Trip Duration (Seconds)')

plt.show()
```

### Observation

Customers generally take longer and more variable trips than subscribers. This indicates that user type has a stronger influence on trip duration than rider age, with subscribers showing more consistent riding patterns.
